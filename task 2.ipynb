```python
import pandas as pd

df = pd.read_excel('/content/data set.xlsx')
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt

sales = df.groupby('Product line')['Total'].sum()

colors = ['red', 'green', 'orange', 'purple', 'pink', 'brown']

sales.plot(kind='bar', figsize=(10,5), color=colors)

plt.title('Sales by Product Line')
plt.xlabel('Product Line')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)

plt.show()
```

```python
df['Date'] = pd.to_datetime(df['Date'])

df['Month'] = df['Date'].dt.month_name()

monthly_sales = df.groupby('Month')['Total'].sum()

monthly_sales = monthly_sales.reindex([
    'January', 'February', 'March'
])

monthly_sales.plot(
    kind='line',
    marker='o',
    color='green',
    figsize=(10,5)
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.grid(True)
plt.show()
```

```python
city_sales = df.groupby('City')['Total'].sum()

colors = ['red', 'green', 'orange']

city_sales.plot(
    kind='bar',
    figsize=(8,5),
    color=colors
)

plt.title('Sales by City')
plt.xlabel('City')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
payment = df['Payment'].value_counts()

colors = ['orange', 'purple', 'green']

payment.plot(
    kind='pie',
    autopct='%1.1f%%',
    colors=colors,
    figsize=(7,7)
)

plt.title('Payment Method Distribution')
plt.ylabel('')
plt.show()
```

```python
total_sales = df['Total'].sum()

total_quantity = df['Quantity'].sum()

average_sales = df['Total'].mean()

total_transactions = len(df)

print("SALES SUMMARY")
print("----------------------")
print("Total Sales:", round(total_sales, 2))
print("Total Quantity:", total_quantity)
print("Average Sales:", round(average_sales, 2))
print("Total Transactions:", total_transactions)
```
